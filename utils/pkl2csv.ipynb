```python
import tkinter as tk
from tkinter import filedialog
import pandas as pd
from pathlib import Path

root = tk.Tk()
root.withdraw()  # hide small tk-window
root.attributes('-topmost', True)  # make sure all tk windows appear in front of other windows

pkl_path = Path(filedialog.askopenfilename(filetypes=[("Pickle files", "*.pkl")]))
if pkl_path.is_file():
    pd.read_pickle(pkl_path).to_csv(pkl_path.with_suffix(".csv"))
```
