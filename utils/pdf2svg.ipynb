# PDF2SVG Converter
Use this notebook to convert PDF files to SVG files. Within the development process of REVOL-E-TION this is used to convert the ```structure_diagram.pdf``` resulting from compiling ```structure_diagram.tex``` to ```structure_diagram.svg``` which then is used in ```README.md```.

The function ```convert_pdf_to_svg()``` accepts a pathlib.Path object to a PDF file which then is converted to one SVG file per page. If the PDF file consists of only one page, the resulting SVG file is stored in the same directory and under the same name as the PDF file (```<pdf_name>.svg```). If the PDF contains more than one page, the SVG files are named ```<pdf_name>_<page_number>.svg``` and are saved to a newly created directory <pdf_name> at the PDF's location.

````python
import fitz  # this imports pymupdf
from pathlib import Path
````

````python
def convert_pdf_to_svg(pdf_path: Path):
    if not pdf_path.is_file():
        raise FileNotFoundError(pdf_path)

    doc = fitz.open(pdf_path)

    if len(doc) == 1:
        onepager = True
        path_output = pdf_path.parent
    else:
        onepager = False
        path_output = pdf_path.parent / path_pdf.stem
        path_output.mkdir(parents=False, exist_ok=True)

    for page_num in range(len(doc)):
        page = doc.load_page(page_num)
        svg = page.get_svg_image()  # returns SVG as string

        svg_filename = f"{pdf_path.stem}{f'_{page_num + 1}' if not onepager else ''}.svg"
        with open((path_output / svg_filename), "w", encoding="utf-8") as f:
            f.write(svg)

    doc.close()
````

````python
path_pdf = Path.cwd().parent / "images" / "structure_diagram.pdf"
convert_pdf_to_svg(path_pdf)
````

### Manual file selection

````python
import tkinter as tk
from tkinter import filedialog

# Create the main Tkinter window
root = tk.Tk()
root.lift()  # Raise the window to the top
root.attributes("-topmost", 1)  # Keep the window on top
root.after(100, lambda: root.attributes("-topmost", 0))  # Remove topmost after 100ms
root.withdraw()

# Open file dialog to select a document
try:
    path_pdf = Path(filedialog.askopenfilename(title="Select a document", filetypes=[("PDF Files", "*.pdf")]))
    root.quit()  # Quit the main loop
    root.destroy()  # Destroy the root window completely

    convert_pdf_to_svg(path_pdf)

except FileNotFoundError:
    print("File not found")
````
