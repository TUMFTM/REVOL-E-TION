# Run REVOL-E-TION example scenarios

```python
# REVOL-E-TION uses plotly for plotting, set the default template to "plotly_white" regardless of the current notebook's theme
import plotly.io as pio

import revoletion

pio.templates.default = "plotly_white"
```

```python
PATH_TO_SCENARIO_FILE = "scenarios.csv"

# Create a new SimulationSettings object with default settings
# SimulationSetting's arguments are the same as the long form of command line arguments:
# solver, n_processes, largescale, debugmode, rerun, rerun_infeasible, key_solcast_api
settings = revoletion.SimulationSettings()

# Create a new SimulationPaths object specifying the path to the scenarios file.
# SimulationPaths' arguments are the same as the long form of command line arguments, also providing the same logic for missing arguments:
# scenario, input, output
paths = revoletion.SimulationPaths.from_plain_paths(scenario=PATH_TO_SCENARIO_FILE)

# Configure the root logger
revoletion.configure_root_logger(paths.log)

# Create and execute the SimulationRun
simulation_run = revoletion.SimulationRun(
    paths=paths,
    settings=settings,  # optional, defaults to revoletion.SimulationSettings() if not provided
)
simulation_run.execute()
```
